```python
import sys, os

def find_project_root(marker='requirements.txt'):
    path = os.getcwd()
    while path != os.path.dirname(path):
        if os.path.exists(os.path.join(path, marker)):
            return path
        path = os.path.dirname(path)
    raise FileNotFoundError(f"Could not find project root (looked for {marker})")

project_root = find_project_root()
if project_root not in sys.path:
    sys.path.append(project_root)

print(project_root)
```

```python
Q1 = df['Total Spent'].quantile(0.25)
Q3 = df['Total Spent'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

outliers = df[df['Total Spent'] > upper_bound]
print(f"Number of statistical outliers in Total Spent: {len(outliers)}")

display(outliers.head(10))
```

```python
discounted_rows = df[
    (df['Discount Applied'] == True) & 
    df['Quantity'].notna() & 
    df['Price Per Unit'].notna() & 
    df['Total Spent'].notna()
]

expected = discounted_rows['Quantity'] * discounted_rows['Price Per Unit']
ratios = (discounted_rows['Total Spent'] / expected).round(2)

print("Discount Multipliers Distribution:")
print(ratios.value_counts())
```

```python
complete_financials = df[
    df['Quantity'].notna() & 
    df['Price Per Unit'].notna() & 
    df['Total Spent'].notna()
]

expected_all = complete_financials['Quantity'] * complete_financials['Price Per Unit']
ratios_all = (complete_financials['Total Spent'] / expected_all).round(2)

print("Multipliers across ALL complete rows (True, False, and Null):")
print(ratios_all.value_counts())
```

```python
df_check = df.dropna(
    subset=[
        'Price Per Unit',
        'Quantity',
        'Total Spent',
        'Discount Applied'
    ]
).copy()

df_check['Expected_Total'] = (
    df_check['Price Per Unit'] *
    df_check['Quantity']
)

df_check['Discount_Amount'] = (
    df_check['Expected_Total'] -
    df_check['Total Spent']
)

df_check.groupby('Discount Applied')['Discount_Amount'].agg(
    ['count', 'min', 'max', 'mean', 'median']
)
```

```python
nan_rows = df[df['Discount Applied'].isna()].copy()

known = df.dropna(
    subset=[
        'Item',
        'Category',
        'Price Per Unit',
        'Discount Applied'
    ]
)

pattern = (
    known
    .groupby(['Item', 'Category', 'Price Per Unit'])['Discount Applied']
    .agg(
        count='count',
        true_rate='mean'
    )
    .reset_index()
)

nan_analysis = nan_rows.merge(
    pattern,
    on=['Item', 'Category', 'Price Per Unit'],
    how='left'
)

nan_analysis[
    [
        'Item',
        'Category',
        'Price Per Unit',
        'count',
        'true_rate'
    ]
].head(30)
```

```python
nan_analysis['inference'] = np.select(
    [
        (nan_analysis['count'] >= 5) & 
        (nan_analysis['true_rate'] >= 0.80),

        (nan_analysis['count'] >= 5) & 
        (nan_analysis['true_rate'] <= 0.20)
    ],
    [
        'True',
        'False'
    ],
    default='Unknown'
)

print(nan_analysis['inference'].value_counts())
```

```python
known = df.dropna(subset=['Item', 'Category', 'Price Per Unit', 'Discount Applied']).copy()
known['Discount Applied'] = known['Discount Applied'].astype(bool)   # هاد السطر يحل المشكلة

pattern_full = known.groupby(['Item', 'Category', 'Price Per Unit'])['Discount Applied'] \
    .agg(count='count', true_rate='mean')

pattern_full['true_rate'] = pattern_full['true_rate'].astype(float)  # تأكيد إضافي

print(pattern_full[pattern_full['count'] >= 5]['true_rate'].describe())
print(pattern_full[pattern_full['count'] >= 5]['true_rate'].round(1).value_counts().sort_index())
```

```python
# Rules To clean the dataset : 
# 1- Columns Name should be in Snake Case way Transaction_ID, ....

# ### Data Type Rules

# - Transaction_ID → string
# - Customer_ID → string
# - Category → string/category
# - Item → string
# - Price_Per_Unit → float64
# - Quantity → integer
# - Total_Spent → float64
# - Payment_Method → string/category
# - Location → string/category
# - Transaction_Date → datetime64
# - Discount_Applied → boolean/NAN

# ### Missing Values Rules
# - Item : HashMap (Category, Price_Per_Unit) 
# - Price_Per_Unit : From Equation (Total_Spent / Quantity)
# - Quantity : From Equation (Total_Spent / Price_Per_Unit)
# - Total_Spent : From Equation (Quantity * Price_Per_Unit)
# -   Discount_Applied:
      # - Convert to nullable boolean (pandas 'boolean' dtype).
      # - Keep missing values as <NA> (unknown) — do not impute.
      # - Verified: discount status is randomly distributed per item 
      #   (mean true_rate ≈ 0.50, normal distribution around 0.5 across 
      #   197 items), confirming no reliable signal exists to infer 
      #   missing values from item, category, or price patterns.
      # - Confirmed the column has zero measurable effect on Price Per 
      #   Unit or Total Spent throughout the dataset.
# - Irrecoverable Rows Rule: Rows with two or more missing values among Price_Per_Unit, Quantity, and Total_Spent are considered irrecoverable and will be removed.
# if we had only quantity we can calculate the Price_Per_Unit and Total_Spent (we don't delete it)

# ### Outlier Rules

# - Detect statistical outliers using IQR.
# - Do not automatically remove outliers.
# - Retain outliers if they satisfy the business rules and valid value ranges.

# Validations :
# Column names / Data types -> should corrected directly 
# Missing values / Irrecoverable rows
# Total_Spent consistency (and check all of the values are not negative or quantity is integer)
# Value ranges / Categorical values
# validate_column_names, validate_data_types, validate_missing_values, validate_irrecoverable_rows,  validate_total_spent_consistency
# ,validate_value_ranges, validate_categorical_values, report_outliers_iqr(Price_Per_Unit), report_outliers_iqr(Total_Spent), validate_duplicates()
# validate_transaction_id_uniqueness(), validate_dates(), validate_item_mapping()

# Note : If two or more of the three financial columns are missing, the row is irrecoverable and will be removed.
# Note : (Category, Price_Per_Unit) → exactly one Item
```
